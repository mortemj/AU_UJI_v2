## 🎯 ¿Qué hace este notebook?

Regenera **dos** ficheros HTML del proyecto leyendo el estado actual del disco:

1. **`docs/html/index.html`** — portada principal con KPIs dinámicos
   (período, registros, variables, tasa abandono, AUC)
2. **`docs/html/orquestador_resumen.html`** — resumen del estado del proyecto
   (modelo ganador, dataset, semáforos de fases, enlaces app)

**No ejecuta ninguna fase.** Solo lee ficheros existentes y genera HTML.

## 📋 ¿Cuándo usarlo?

- El profesor o tribunal abre el proyecto por primera vez y quiere ver
  el estado sin lanzar el orquestador maestro entero
- Tras corregir manualmente algún notebook y querer regenerar el HTML
- Tras tocar `metricas_modelo.json` o cualquier parquet y refrescar visualización
- Como verificación rápida tras un `git pull`

## ⚠️ Requisitos

- `src/index_generator.py` con `actualizar_index()`
- `src/html/generar_resumen_proyecto.py` con `generar_resumen_orquestador()`
- Entorno: `tfm_abandono`

## 📝 Comportamiento si faltan datos

Si algún fichero esperado no existe (por ejemplo `metricas_modelo.json` cuando
F6 aún no se ha ejecutado), el HTML se genera **igualmente** con un cartel
"Pendiente — ejecuta la Fase X". Nunca se inventan valores.

## ➡️ Salida

Dos ficheros actualizados en `docs/html/`. Tras ejecutar este notebook, abrir:

- `docs/html/index.html` (portada)
- `docs/html/orquestador_resumen.html` (resumen del proyecto)


```python
# ============================================================================
# CELDA 1 — CONFIGURACIÓN DE RUTAS (ROOT robusto)
# ============================================================================
import sys
from pathlib import Path

def _encontrar_root(start: Path) -> Path:
    """Sube por los padres hasta encontrar la carpeta src/."""
    for parent in [start] + list(start.parents):
        if (parent / 'src').is_dir():
            return parent
    raise FileNotFoundError(f'No se encontró src/ subiendo desde {start}')

ROOT = _encontrar_root(Path.cwd())
sys.path.insert(0, str(ROOT))

print(f'ROOT: {ROOT}')
```

```python
# ============================================================================
# CELDA 2 — IMPORTS
# ============================================================================
from src.index_generator import actualizar_index
from src.html.generar_resumen_proyecto import generar_resumen_orquestador

print('✓ Imports OK')
```

```python
# ============================================================================
# CELDA 3 — REGENERAR index.html (portada principal)
# ============================================================================
# Lee dinámicamente Excel originales, parquets y metricas_modelo.json.
# Si algún dato falta, los KPIs aparecen translúcidos con "Pendiente".
# ============================================================================

ruta_index = actualizar_index(verbose=True)
print(f'\n✅ Index regenerado: {ruta_index}')
```

```python
# ============================================================================
# CELDA 4 — REGENERAR orquestador_resumen.html
# ============================================================================
# Sin datos de ejecución (ya que no hemos lanzado el orquestador en esta
# sesión), el HTML mostrará "Sin ejecución registrada en esta sesión" en el
# bloque correspondiente. Los demás bloques (modelo, dataset, semáforos,
# app) se rellenan dinámicamente con los datos reales del proyecto.
# ============================================================================

ruta_resumen = generar_resumen_orquestador(
    resultados=None,         # No hay ejecución que reportar
    t_inicio_global=None,
    parado=False,
    verbose=True,
)
print(f'\n✅ Resumen regenerado: {ruta_resumen}')
```

```python
# ============================================================================
# CELDA 5 — RESUMEN
# ============================================================================
print('=' * 60)
print('REGENERACIÓN HTML COMPLETADA')
print('=' * 60)
print()
print('Ficheros actualizados:')
print(f'  ✓ {ruta_index}')
print(f'  ✓ {ruta_resumen}')
print()
print('📌 Siguiente: abrir cualquiera de los dos en el navegador')
print('   o visitar https://mortemj.github.io/AU_UJI_v2/')
```
