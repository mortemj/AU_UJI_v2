```python
import pandas as pd
from pathlib import Path

# Detectar ruta automáticamente — buscar el dataset
rutas_candidatas = [
    Path('data/02_processed/df_alumno.parquet'),
    Path('../data/02_processed/df_alumno.parquet'),
    Path('../../data/02_processed/df_alumno.parquet'),
]

ruta = None
for r in rutas_candidatas:
    if r.exists():
        ruta = r.resolve()
        break

if ruta is None:
    print('⚠️ No encontrado. Pon la ruta manualmente:')
    print('   ruta = Path(r"C:\\ruta\\completa\\df_alumno.parquet")')
else:
    print(f'✅ Encontrado: {ruta}')
    print()
    df = pd.read_parquet(ruta)
    print(f'📐 Shape: {df.shape}')
    print(f'   Filas:    {df.shape[0]:,}')
    print(f'   Columnas: {df.shape[1]}')
    print()
    print('📋 Lista de columnas:')
    for i, c in enumerate(df.columns, 1):
        print(f'   {i:2d}. {c}')
    print()
    if 'per_id_ficticio' in df.columns:
        n_alumnos = df['per_id_ficticio'].nunique()
        print(f'👥 Alumnos únicos (per_id_ficticio): {n_alumnos:,}')
    if 'titulacion' in df.columns:
        n_tit = df['titulacion'].nunique()
        print(f'🎓 Titulaciones únicas: {n_tit}')
```
