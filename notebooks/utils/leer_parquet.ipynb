# 🔍 leer_parquet.ipynb

**Notebook simple para abrir y explorar cualquier fichero `.parquet` del proyecto.**

## Cómo se usa

1. Cambia la línea `RUTA = ...` en la celda 2 por el fichero que quieras ver.
2. Pulsa **Run All** (o ejecuta celda a celda).
3. Listo: ves el fichero, sus columnas, ejemplos y estadísticas.

## Qué hace cada celda

- **Celda 1**: imports y ROOT (igual que los demás notebooks).
- **Celda 2**: elige el fichero y lo carga.
- **Celda 3**: te enseña shape, columnas y tipos.
- **Celda 4**: te enseña las primeras 5 filas.
- **Celda 5**: te enseña estadísticas básicas (medias, mínimos, máximos).
- **Celda 6**: te dice si hay nulos y cuántos.

**No modifica el fichero — solo lo lee.**

````python
# ── Celda 1: setup ────────────────────────────────────────────────────────────
# Detectar ROOT del proyecto subiendo niveles hasta encontrar src/
import sys
from pathlib import Path

ROOT = Path.cwd()
for _ in range(6):
    if (ROOT / 'src').exists():
        break
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT))

import pandas as pd

# Mostrar todas las columnas (sin truncar)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

print(f'ROOT: {ROOT}')
print(f'Pandas: {pd.__version__}')
````

````python
# ── Celda 2: elegir fichero parquet ───────────────────────────────────────────
# 👇 CAMBIA AQUÍ el fichero que quieres leer.
# Algunos ejemplos comunes (descomenta el que necesites):

RUTA = ROOT / 'data' / '06_evaluacion' / 'meta_test_app.parquet'
# RUTA = ROOT / 'data' / '06_evaluacion' / 'meta_test.parquet'
# RUTA = ROOT / 'data' / '05_modelado'   / 'X_test_prep.parquet'
# RUTA = ROOT / 'data' / '05_modelado'   / 'X_test.parquet'
# RUTA = ROOT / 'data' / '05_modelado'   / 'y_test.parquet'
# RUTA = ROOT / 'data' / '05_modelado'   / 'results' / 'resultados_maestro.parquet'
# RUTA = ROOT / 'data' / '04_eda'        / 'df_eda_final.parquet'
# RUTA = ROOT / 'data' / '03_features'   / 'df_alumno_limpio.parquet'
# RUTA = ROOT / 'data' / '03_features'   / 'dataset_final_tfm.parquet'

assert RUTA.exists(), f'❌ No encontrado: {RUTA}'
df = pd.read_parquet(RUTA)

print(f'✅ Cargado: {RUTA.name}')
print(f'   Ruta:   {RUTA}')
print(f'   Tamaño: {RUTA.stat().st_size / 1024:.1f} KB')
````

````python
# ── Celda 3: info general ─────────────────────────────────────────────────────
# Te dice cuántas filas, cuántas columnas, y qué tipo es cada columna
print(f'Shape: {df.shape}  ({df.shape[0]:,} filas × {df.shape[1]} columnas)')
print()
print(f'Índice: desde {df.index.min()} hasta {df.index.max()}')
print()
print('Columnas y tipos:')
print(df.dtypes)
````

````python
# ── Celda 4: primeras 5 filas ─────────────────────────────────────────────────
# Para ver cómo es cada fila por dentro
df.head()
````

````python
# ── Celda 5: estadísticas básicas (solo columnas numéricas) ──────────────────
# Te dice min, max, media, etc. de cada columna numérica
df.describe()
````

````python
# ── Celda 6: nulos ────────────────────────────────────────────────────────────
# Te dice cuántos valores faltan en cada columna
nulos = df.isnull().sum()
nulos_reales = nulos[nulos > 0]

if len(nulos_reales) == 0:
    print('✅ Sin nulos en ninguna columna')
else:
    print('Columnas con nulos:')
    print(nulos_reales)
    print()
    print(f'Total filas con al menos un nulo: {df.isnull().any(axis=1).sum():,}')
````
