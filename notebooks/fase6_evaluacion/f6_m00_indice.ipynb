````python
# ============================================================
# CELDA 1: CONFIGURACIÓN DE RUTAS
# Detectar ROOT subiendo niveles hasta encontrar src/
# ============================================================
import sys
import json
from pathlib import Path

def _encontrar_root(start: Path) -> Path:
    for parent in [start] + list(start.parents):
        if (parent / 'src').is_dir():
            return parent
    raise FileNotFoundError('No se encontró el directorio src/ subiendo desde ' + str(start))

ROOT = _encontrar_root(Path.cwd())
sys.path.insert(0, str(ROOT))

# Ahora (v2): leer el ganador dinámicamente desde metricas_modelo.json
RUTA_JSON = ROOT / 'data' / '06_evaluacion' / 'metricas_modelo.json'
assert RUTA_JSON.exists(), f'❌ No encontrado: {RUTA_JSON}'
with open(RUTA_JSON, encoding='utf-8') as f:
    meta_json = json.load(f)

nombre_ganador  = meta_json['modelo_nombre']
familia_ganador = meta_json['modelo_familia']
auc_ganador     = meta_json.get('auc_test', 0.9564)
f1_ganador      = meta_json.get('f1_test', 0.8334)

print(f'ROOT:           {ROOT}')
print(f'Modelo ganador: {nombre_ganador} ({familia_ganador})')
print(f'AUC test:       {auc_ganador:.4f}')
print(f'F1 test:        {f1_ganador:.4f}')
````
